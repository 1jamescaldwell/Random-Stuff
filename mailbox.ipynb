```python

import mailbox
import pandas as pd
from email.utils import parseaddr
from collections import Counter

# ==============================
# CONFIGURATION
# ==============================
mbox_path = r'path here.mbox'
output_path = r'path here.xlsx'

# ==============================
# PROCESS MBOX
# ==============================
mbox = mailbox.mbox(mbox_path)

sender_counts = Counter()

for message in mbox:
    # Extract sender email
    sender = message.get("From", "")

    name, email = parseaddr(sender)

    if email:
        sender_counts[email.lower()] += 1

# ==============================
# CREATE EXCEL FILE
# ==============================
df = pd.DataFrame(
    sender_counts.items(),
    columns=["email", "number of emails"]
)

# Sort largest to smallest
df = df.sort_values(
    by="number of emails",
    ascending=False
).reset_index(drop=True)

# Export to Excel
df.to_excel(output_path, index=False)

print(f"Finished!")
print(f"Total unique senders: {len(df):,}")
print(f"Total emails processed: {sum(sender_counts.values()):,}")
# print(f"Excel saved to: {output_path}")
```
